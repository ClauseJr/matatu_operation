```python
df['vehicle_capacity'] = (
    df['capacity']
    .astype(str)
    .str.extract(r'(-?\d+\.?\d*)')[0]
    .astype(int)
)

df['distance(km)'] = (
    df['distance_km']
    .astype(str)
    .str.extract(r'(-?\d+\.?\d*)')[0]
    .astype(float)
)

df['fare(ksh)'] = (
    df['fare_kes']
    .astype(str)
    .str.extract(r'(-?\d+\.?\d*)')[0]
    .astype(float)
)


df['fuel_consumption(Litrers/100km)'] = (
    df['fuel_consumption']
    .astype(str)
    .str.replace('100km', '', regex = False)
    .str.extract(r'(-?\d+\.?\d*)')[0]
    .astype(float)
)

df['fuel_price(unit)'] = (
    df['fuel_price_per_unit']
    .astype(str)
    .str.extract(r'(-?\d+\.?\d*)')[0]
    .astype(float)
)


print(df.to_string(max_rows = 10))
```

```python
df = df.drop(columns = ['capacity', 'distance_km', 'fare_kes', 'revenue_kes', 'fuel_consumption', 'fuel_price_per_unit'])
print(df.to_string(max_rows = 10))
```

```python
df['revenue(ksh)'] = df['passenger_count'] * df['fare(ksh)']

df['fuel_cost'] = ((df['fuel_consumption(Litrers/100km)'] * df['distance(km)'] / 100) * df['fuel_price(unit)']).round(2)

df['profit(ksh)'] = df['revenue(ksh)'] - df['fuel_cost'].round(2)
```

```python
print(df.to_string(max_rows = 10))
```

```python
def parse_messy_date(val):
    val = str(val).strip()
    
    if val.isdigit() and 40000 <= int(val) <= 50000:
        return pd.to_datetime(int(val), unit='D', origin='1899-12-30')

    if val.isdigit() and len(val) >= 9:
        return pd.to_datetime(int(val), unit='s')
    
    return pd.to_datetime(val, format='mixed', errors='coerce')

df['date_clean'] = df['date_raw'].apply(parse_messy_date)

print('Unparsed:', df['date_clean'].isna().sum())
print(df.loc[df['date_clean'].isna(), 'date_raw'].unique())
```

```python
def winsorize_by_route(group):
    group = group.copy()

    group["distance(km)"] = group["distance(km)"].abs()
    
    Q1 = group["distance(km)"].quantile(0.25)
    Q3 = group["distance(km)"].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    group["distance_km_clean"] = group["distance(km)"].clip(
        lower=lower_bound,
        upper=upper_bound
    )
    return group

route_numbers = df["route_number"].copy()

df = df.groupby("route_number", group_keys=False).apply(
    winsorize_by_route
).reset_index(drop=True)

if "route_number" not in df.columns:
    df["route_number"] = route_numbers.values
    print("route_number restored")

print(df.columns.tolist())
```

```python

print(df.index)
print(df.columns.tolist())


route_check = df.groupby("route_number").agg(
    original_min   = ("distance(km)",      "min"),
    original_max   = ("distance(km)",      "max"),
    original_mean  = ("distance(km)",      "mean"),
    clean_min      = ("distance_km_clean", "min"),
    clean_max      = ("distance_km_clean", "max"),
    clean_mean     = ("distance_km_clean", "mean"),
    rows_corrected = ("distance(km)", lambda x: (        # ← original column
        (x < x.quantile(0.25) - 1.5 * (x.quantile(0.75) - x.quantile(0.25))) |
        (x > x.quantile(0.75) + 1.5 * (x.quantile(0.75) - x.quantile(0.25)))
    ).sum())
).round(2)

print(route_check.to_string())
print(f"\nTotal rows: {len(df)} (unchanged)")
```
